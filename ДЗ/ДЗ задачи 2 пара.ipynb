### Задача 1.38
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

### Задача 1.40
> Из 20 филиалов Сбербанка 10 расположены за чертой города. Для обследования случайным образом отобрано 5 филиалов. Какова вероятность того, что среди отобранных окажется в черте города: а) 3 филиала; б) хотя бы один?

* а) $P(A) = \frac{C_{10}^3 \cdot C_{10}^2}{C_{20}^5}$

```python
round(sc.comb(10, 3) * sc.comb(10, 2) / sc.comb(20, 5), 4)
```

* б) $P(B) = 1 - \frac{C_{10}^5}{C_{20}^5}$

```python
round(1 - sc.comb(10, 5) / sc.comb(20, 5), 4)
```

### Задача 1.43

```python
round(9 / (9 * 10**4), 5)
```

* в) $P(C) = \frac{5^5}{9 \cdot 10^4}$

```python
round(5 * (5**4) / (9 * 10**4), 4)
```

### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

```python
m = 0
n = 0

for q in range(1, 7):
    for w in range(1, 7):
        for e in range(1, 7):
            m += 1
            if q + w + e == 11:
                n += 1

round(n / m, 4)
```

### Задача 1.50
> Два лица условились встретиться в определенном месте между 18 и 19 ч и договорились, что пришедший первым ждет другого в течение 15 мин., после чего уходит. Найти вероятность их встречи, если приход каждого в течение указанного часа может произойти в любое время и моменты прихода независимы.

```python
from scipy.integrate import quad

def f1(x):
    return min(x + 15/60, 1)

def f2(x):
    return max(x - 15/60, 0)

r1 = quad(f1, 0, 1)
r2 = quad(f2, 0, 1)

round(r1[0] - r2[0], 4)
```

### Задача 1.51
> Какова вероятность того, что наудачу брошенная в круг точка окажется внутри вписанного в него квадрата?

* $P(A) = \frac{S_{кв}}{S_{кр}} = \frac{2R^2}{\pi R^2} = \frac{2}{\pi}$

```python
round(2 / math.pi, 4)
```
