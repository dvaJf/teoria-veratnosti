### Задача 2.9
> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов.


* 1) $P_9(4) = C_9^4 \cdot 0,2^4 \cdot 0,8^5$


```python
round(stats.binom.pmf(4, 9, 0.2), 4)
```

* 2а) $P_9(m < 2) = P_9(0) + P_9(1)$


```python
round(stats.binom.cdf(1, 9, 0.2), 4)
```

* 2б) $P_9(m \le 2) = P_9(0) + P_9(1) + P_9(2)$


```python
round(stats.binom.cdf(2, 9, 0.2), 4)
```

* 2в) $P_9(m \ge 2) = 1 - P_9(m < 2)$


```python
round(1 - stats.binom.cdf(1, 9, 0.2), 4)
```

* 2г) $np - q \le m_0 \le np + p \implies 1 \le m_0 \le 2 \implies m_0 = 1, m_0 = 2$
$P_{\text{наивер}} = P_9(1) + P_9(2)$


```python
round(stats.binom.pmf(1, 9, 0.2) + stats.binom.pmf(2, 9, 0.2), 4)
```

### Задача 2.10
> Завод отправил на базу 10 000 стандартных изделий. Среднее число изделий, повреждаемых при транспортировке, составляет 0,02%. Найти вероятность того, что из 10 000 изделий: 1) будет повреждено: а) 3; б) по крайней мере 3; 2) не будет повреждено: а) 9997; б) хотя бы 9997.


* 1а) $\lambda = np = 10000 \cdot 0,0002 = 2$
$P(m = 3) \approx \frac{2^3 e^{-2}}{3!}$


```python
round(stats.poisson.pmf(3, 2), 4)
```

* 1б) $P(m \ge 3) = 1 - P(m \le 2)$


```python
round(1 - stats.poisson.cdf(2, 2), 4)
```

* 2а) $m = 10000 - 9997 = 3$


```python
round(stats.poisson.pmf(3, 2), 4)
```

* 2б) $m \le 3$


```python
round(stats.poisson.cdf(3, 2), 4)
```

### Задача 2.11
> По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины: а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520.


* а) $n = 1000, p = 0,5, q = 0,5, npq = 250$
$x = \frac{480 - 500}{\sqrt{250}} \approx -1,265$
$P_{1000}(480) \approx \frac{\varphi(x)}{\sqrt{npq}}$


```python
n = 1000
p = 0.5
s = math.sqrt(n * p * (1 - p))
x = (480 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

* б) $np - q \le m_0 \le np + p \implies m_0 = 500$


```python
x = (500 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

* в) $P(480 \le m \le 1000) = \Phi(x_2) - \Phi(x_1)$


```python
x1 = (480 - n * p) / s
x2 = (1000 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```

* г) $P(480 \le m \le 520) = \Phi(x_2) - \Phi(x_1)$


```python
x1 = (480 - n * p) / s
x2 = (520 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```
