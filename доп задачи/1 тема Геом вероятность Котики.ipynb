Два котика хотят есть случайным образом с 08:00 до 22:00 (ночью они бегают по квартире и никогда не хотят кушать). Хозяин дома с 08:00–10:00, 13:00–14:00, 18:00–22:00. Какова вероятность, что оба котика будут покормлены? Если котика сразу не покормить он обидется и день будет испорчен

```python
import matplotlib.pyplot as plt
import random
```

## Симуляция

```python
N = 100000

c = 0

x_true = []
y_true = []
x_false = []
y_false = []

for i in range(N):
    cat1 = random.uniform(8, 22)
    cat2 = random.uniform(8, 22)
    
    flag1 = False
    if (8 <= cat1 <= 10) or (13 <= cat1 <= 14) or (18 <= cat1 <= 22):
        flag1 = True
        
    flag2 = False
    if (8 <= cat2 <= 10) or (13 <= cat2 <= 14) or (18 <= cat2 <= 22):
        flag2 = True
        
    if flag1 and flag2:
        c += 1
        x_true.append(cat1)
        y_true.append(cat2)
    else:
        x_false.append(cat1)
        y_false.append(cat2)

res = c / N

res_true = (0.5) ** 2

print("Опыт:", res)
print("Теоретическая вероятность: ", res_true)
```

## Визуализация

```python
plt.figure(figsize=(8, 8))

plt.scatter(x_false, y_false, color='red', s=5)
plt.scatter(x_true, y_true, color='green', s=5)

plt.xlim(8, 22)
plt.ylim(8, 22)
plt.xticks([8, 10, 13, 14, 18, 22])
plt.yticks([8, 10, 13, 14, 18, 22])


plt.title("Когда котики хотели есть")
plt.xlabel("Время первого котика")
plt.ylabel("Время второго котика")

plt.show()
```
