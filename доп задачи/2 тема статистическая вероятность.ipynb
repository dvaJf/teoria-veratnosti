## Кот гуляет по клавиатуре и случайно нажимает 3 буквы подряд. Какова вероятностьб что кот наберёт ровно слово "cat"?

```python
import matplotlib.pyplot as plt
import random
```

## Симуляция

```python
a = list('abcdefghijklmnopqrstuvwxyz') 
w = ['c', 'a', 't']

N = 10**8
c = 0

for i in range(N):
    p = random.choices(a, k=3)

    flag = False
    if p == w:
        flag = True

    if flag:
        c += 1

r = c / N

res = (1 / 26) ** 3

print("Опыт", r)
print("Теоретическая вероятность ", res)
```

## Визуализация

```python
n = list(range(10, 50_000_001, 100_000))
r = []

h = 0
c = 0

for n1 in n:
    while c < n1:
        p = random.choices(a, k=3)
        if p == w:
            h += 1
        c += 1
    r.append(h / c)

plt.figure(figsize=(8, 5))

plt.plot(n, r, color='green', label='Частотная вероятность')
plt.axhline(y=res, color='red', linestyle='--', label='Классическая вероятность')

plt.title("Сходимость частоты к вероятности")
plt.xlabel("Число опытов")
plt.ylabel("P(A)")
plt.legend()

plt.show()
```
